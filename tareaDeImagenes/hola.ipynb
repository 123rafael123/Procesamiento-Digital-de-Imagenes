```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('pgn.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].imshow(img_rgb)
axs[0].set_title('Imagen Original')

fig_img=axs[1].imshow(img_rgb[...,0], cmap='gray')
fig.colorbar(fig_img, ax=axs[1], shrink=0.5)
axs[1].set_title('Canal Rojo')

fig_img=axs[2].imshow(img_rgb[...,1], cmap='gray')
fig.colorbar(fig_img, ax=axs[2], shrink=0.5)
axs[2].set_title('Canal Verde')

fig_img=axs[3].imshow(img_rgb[...,2], cmap='gray')
fig.colorbar(fig_img, ax=axs[3], shrink=0.5)
axs[3].set_title('Canal Azul')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread("pego.png")
imagen = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

alto, ancho, _ = imagen.shape
mitad_alto = alto // 2
mitad_ancho = ancho // 2

resultado = np.zeros_like(imagen)

resultado[0:mitad_alto, 0:mitad_ancho] = imagen[0:mitad_alto, 0:mitad_ancho] * [1,0,0]
resultado[0:mitad_alto, mitad_ancho:ancho] = imagen[0:mitad_alto, mitad_ancho:ancho] * [0,1,0]
resultado[mitad_alto:alto, 0:mitad_ancho] = imagen[mitad_alto:alto, 0:mitad_ancho] * [0,0,1]
resultado[mitad_alto:alto, mitad_ancho:ancho] = imagen[mitad_alto:alto, mitad_ancho:ancho] * [1,1,0]

plt.imshow(resultado)
plt.axis("off")
plt.show()
```
