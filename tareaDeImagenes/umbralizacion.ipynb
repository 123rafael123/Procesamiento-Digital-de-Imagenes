umbralizacion binaria y binaria inversa

```python
import cv2
import matplotlib.pyplot as plt
imagen=cv2.imread('pego.png',0)#lee en escala de grises
#umbralizacion binaria
_,umbral_binario=cv2.threshold(imagen,200,255,cv2.THRESH_BINARY)
_,umbral_binaria_inv=cv2.threshold(imagen,127,255,cv2.THRESH_BINARY)

plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(imagen,cmap='gray')

plt.subplot(1,3,2)
plt.imshow(umbral_binario,cmap='gray')

plt.subplot(1,3,3)
plt.imshow(umbral_binaria_inv,cmap='gray')
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# leer imagen en escala de grises (simulando sensor Bayer)
imagen = cv2.imread('pego.png',0)

# aplicar diferentes filtros Bayer
bayer_bg = cv2.cvtColor(imagen, cv2.COLOR_BAYER_BG2BGR)
bayer_gb = cv2.cvtColor(imagen, cv2.COLOR_BAYER_GB2BGR)
bayer_rg = cv2.cvtColor(imagen, cv2.COLOR_BAYER_RG2BGR)
bayer_gr = cv2.cvtColor(imagen, cv2.COLOR_BAYER_GR2BGR)

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.title("Bayer BG")
plt.imshow(cv2.cvtColor(bayer_bg, cv2.COLOR_BGR2RGB))

plt.subplot(2,2,2)
plt.title("Bayer GB")
plt.imshow(cv2.cvtColor(bayer_gb, cv2.COLOR_BGR2RGB))

plt.subplot(2,2,3)
plt.title("Bayer RG")
plt.imshow(cv2.cvtColor(bayer_rg, cv2.COLOR_BGR2RGB))

plt.subplot(2,2,4)
plt.title("Bayer GR")
plt.imshow(cv2.cvtColor(bayer_gr, cv2.COLOR_BGR2RGB))

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

imagen = cv2.imread('pego.png')

rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

yuv = cv2.cvtColor(rgb, cv2.COLOR_RGB2YUV)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("RGB")
plt.imshow(rgb)

plt.subplot(1,2,2)
plt.title("YUV")
plt.imshow(yuv)

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

imagen = cv2.imread('pego.png')

rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

yuv = cv2.cvtColor(rgb, cv2.COLOR_RGB2YUV)

rgb_convertido = cv2.cvtColor(yuv, cv2.COLOR_YUV2RGB)

plt.figure(figsize=(10,5))

plt.subplot(1,3,1)
plt.title("Original RGB")
plt.imshow(rgb)

plt.subplot(1,3,2)
plt.title("YUV")
plt.imshow(yuv)

plt.subplot(1,3,3)
plt.title("YUV → RGB")
plt.imshow(rgb_convertido)

plt.show()
```
