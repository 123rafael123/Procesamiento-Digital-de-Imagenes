```python
import cv2
import matplotlib.pyplot as plt
imagen=cv2.imread('pego.png',0)#lee en escala de grises
#umbralizacion binaria
_,umbral_binario=cv2.threshold(imagen,127,255,cv2.THRESH_BINARY)
_,umbral_binaria_inv=cv2.threshold(imagen,127,255,cv2.THRESH_BINARY)

plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(imagen,cmap='gray')

plt.subplot(1,3,2)
plt.imshow(umbral_binario,cmap='gray')

plt.subplot(1,3,3)
plt.imshow(umbral_binaria_inv,cmap='gray')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('pego.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
h, w, _ = img_rgb.shape

def mosaico_color_predominante(img, patron, predominante):
    mosaico = np.zeros_like(img)
    for i in range(h):
        for j in range(w):
            r, g, b = img[i,j]

            if patron=="RGGB":
                if predominante=="R":  
                    if i%2==0 and j%2==0 or i%2==1 and j%2==0:
                        mosaico[i,j,0] = r
                    elif i%2==0 and j%2==1:
                        mosaico[i,j,1] = g
                    else:
                        mosaico[i,j,2] = b
                elif predominante=="B":  
                    if i%2==1 and j%2==1 or i%2==0 and j%2==0:
                        mosaico[i,j,2] = b
                    elif i%2==0 and j%2==1:
                        mosaico[i,j,1] = g
                    else:
                        mosaico[i,j,0] = r
                else: 
                    if i%2==0 and j%2==1 or i%2==1 and j%2==0:
                        mosaico[i,j,1] = g
                    elif i%2==0 and j%2==0:
                        mosaico[i,j,0] = r
                    else:
                        mosaico[i,j,2] = b

            elif patron=="BGGR":
                if predominante=="R":
                    if i%2==1 and j%2==1 or i%2==0 and j%2==1:
                        mosaico[i,j,0] = r
                    elif i%2==0 and j%2==0:
                        mosaico[i,j,2] = b
                    else:
                        mosaico[i,j,1] = g
                elif predominante=="B":
                    if i%2==0 and j%2==0 or i%2==1 and j%2==0:
                        mosaico[i,j,2] = b
                    elif i%2==0 and j%2==1:
                        mosaico[i,j,1] = g
                    else:
                        mosaico[i,j,0] = r
                else:
                    if i%2==0 and j%2==1 or i%2==1 and j%2==0:
                        mosaico[i,j,1] = g
                    elif i%2==0 and j%2==0:
                        mosaico[i,j,2] = b
                    else:
                        mosaico[i,j,0] = r

            elif patron=="GBRG":
                if predominante=="R":
                    if i%2==1 and j%2==0 or i%2==0 and j%2==1:
                        mosaico[i,j,0] = r
                    elif i%2==0 and j%2==0:
                        mosaico[i,j,2] = b
                    else:
                        mosaico[i,j,1] = g
                elif predominante=="B":
                    if i%2==0 and j%2==1 or i%2==1 and j%2==1:
                        mosaico[i,j,2] = b
                    elif i%2==0 and j%2==0:
                        mosaico[i,j,1] = g
                    else:
                        mosaico[i,j,0] = r
                else:
                    if i%2==0 and j%2==0 or i%2==1 and j%2==1:
                        mosaico[i,j,1] = g
                    elif i%2==0 and j%2==1:
                        mosaico[i,j,2] = b
                    else:
                        mosaico[i,j,0] = r

            elif patron=="GRBG":
                if predominante=="R":
                    if i%2==0 and j%2==1 or i%2==1 and j%2==0:
                        mosaico[i,j,0] = r
                    elif i%2==0 and j%2==0:
                        mosaico[i,j,1] = g
                    else:
                        mosaico[i,j,2] = b
                elif predominante=="B":
                    if i%2==1 and j%2==0 or i%2==1 and j%2==1:
                        mosaico[i,j,2] = b
                    elif i%2==0 and j%2==0:
                        mosaico[i,j,1] = g
                    else:
                        mosaico[i,j,0] = r
                else:
                    if i%2==0 and j%2==0 or i%2==1 and j%2==1:
                        mosaico[i,j,1] = g
                    elif i%2==0 and j%2==1:
                        mosaico[i,j,0] = r
                    else:
                        mosaico[i,j,2] = b
    return mosaico


mosaico_rggb = mosaico_color_predominante(img_rgb,"RGGB","R")
mosaico_bggr = mosaico_color_predominante(img_rgb,"BGGR","B")
mosaico_gbrg = mosaico_color_predominante(img_rgb,"GBRG","G")
mosaico_grbg = mosaico_color_predominante(img_rgb,"GRBG","R")


plt.figure(figsize=(12,6))

plt.subplot(2,2,1)
plt.title("Mosaico RGGB (rojo)")
plt.imshow(mosaico_rggb)
plt.axis('off')

plt.subplot(2,2,2)
plt.title("Mosaico BGGR (azul)")
plt.imshow(mosaico_bggr)
plt.axis('off')

plt.subplot(2,2,3)
plt.title("Mosaico GBRG (verde)")
plt.imshow(mosaico_gbrg)
plt.axis('off')

plt.subplot(2,2,4)
plt.title("Mosaico GRBG (rojo)")
plt.imshow(mosaico_grbg)
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

imagen = cv2.imread('pego.png')

rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

yuv = cv2.cvtColor(rgb, cv2.COLOR_RGB2YUV)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("RGB")
plt.imshow(rgb)

plt.subplot(1,2,2)
plt.title("YUV")
plt.imshow(yuv)

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

imagen = cv2.imread('pego.png')

rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

yuv = cv2.cvtColor(rgb, cv2.COLOR_RGB2YUV)

rgb_convertido = cv2.cvtColor(yuv, cv2.COLOR_YUV2RGB)

plt.figure(figsize=(10,5))

plt.subplot(1,3,1)
plt.title("Original RGB")
plt.imshow(rgb)

plt.subplot(1,3,2)
plt.title("YUV")
plt.imshow(yuv)

plt.subplot(1,3,3)
plt.title("YUV → RGB")
plt.imshow(rgb_convertido)

plt.show()
```
