```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Esta funcion te muestra la foto, si queres en gris o a full color
def mostrar_imagen(imagen, titulo, es_gris=False):
    if es_gris:
        plt.imshow(imagen, cmap='gray')
    else:
        img_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
        plt.imshow(img_rgb)
    plt.title(titulo)
    plt.axis('off')


# BLOQUE 1 — Primera imagen
img1 = cv2.imread('rafa.jpg')
gris1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)

cv2.imwrite('imagen1_color.jpg', img1)
cv2.imwrite('imagen1_gris.jpg', gris1)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
mostrar_imagen(img1, 'Color Original (bien chevere)')

plt.subplot(1, 2, 2)
mostrar_imagen(gris1, 'Escala de Grises (modo clasico)', es_gris=True)
plt.show()

b, g, r = cv2.split(img1)

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(r, cmap='jet')
plt.title('Rojo (tono de mentira)')

plt.subplot(1, 3, 2)
plt.imshow(g, cmap='Greens')
plt.title('Verde (bien vivo)')

plt.subplot(1, 3, 3)
plt.imshow(b, cmap='Blues')
plt.title('Azul (chiquitito)')
plt.show()


# BLOQUE 2 — Segunda imagen
img2 = cv2.imread('rafa.jpg')
gris2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

cv2.imwrite('imagen2_color.jpg', img2)
cv2.imwrite('imagen2_gris.jpg', gris2)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
mostrar_imagen(img2, 'Color Original (bacan)')

plt.subplot(1, 2, 2)
mostrar_imagen(gris2, 'Escala de Grises (modo clasico)', es_gris=True)
plt.show()

b, g, r = cv2.split(img2)

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(r, cmap='jet')
plt.title('Rojo (con onda)')

plt.subplot(1, 3, 2)
plt.imshow(g, cmap='Greens')
plt.title('Verde (bien chevere)')

plt.subplot(1, 3, 3)
plt.imshow(b, cmap='Blues')
plt.title('Azul (tranqui)')
plt.show()


# BLOQUE 3 — Tercera imagen
img3 = cv2.imread('rafa.jpg')
gris3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)

cv2.imwrite('imagen3_color.jpg', img3)
cv2.imwrite('imagen3_gris.jpg', gris3)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
mostrar_imagen(img3, 'Color Original (re piola)')

plt.subplot(1, 2, 2)
mostrar_imagen(gris3, 'Escala de Grises (modo clasico)', es_gris=True)
plt.show()

b, g, r = cv2.split(img3)

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(r, cmap='jet')
plt.title('Rojo (con power)')

plt.subplot(1, 3, 2)
plt.imshow(g, cmap='Greens')
plt.title('Verde (full vida)')

plt.subplot(1, 3, 3)
plt.imshow(b, cmap='Blues')
plt.title('Azul (modo tranquilo)')
plt.show()
```
